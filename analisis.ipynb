# Análisis de Espectrofotometría
Carga de datos y gráfica de absorbancia.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo CSV (separado por tabuladores y con comas como decimales)
df = pd.read_csv('Barrido_02_03_2026_10_33_15.csv', sep='\t', decimal=',', encoding='utf-16le')

# Limpiar las columnas eliminando espacios extras si los hay
df.columns = df.columns.str.strip()

# Mostrar las primeras filas del DataFrame
display(df.head())
```

```python
# Configurar la gráfica
plt.figure(figsize=(12, 6))

# Trazar cada columna excepto la longitud de onda
x = df['Wavelength(nm)']
columns_to_plot = ['TVS 1(Abs)', 'CS 1(Abs)', 'TNS 1(Abs)', 'TVK 1(Abs)', 'CK 1(Abs)', 'TNK 1(Abs)']
colors = ['#d62728', '#1f77b4', '#2ca02c', '#bcbd22', '#e377c2', '#17becf'] # Aproximados a la imagen

for col, color in zip(columns_to_plot, colors):
    if col in df.columns:
        plt.plot(x, df[col], label=col.replace('(Abs)', '').strip(), color=color, linewidth=1.5)

# Añadir las líneas verticales
plt.axvline(x=417, color='darkblue', linestyle='-', label='417 nm', linewidth=1.5)
plt.text(417 + 5, 5, '417', rotation=90, verticalalignment='top', color='black')

plt.axvline(x=873, color='darkgreen', linestyle='-', label='873 nm', linewidth=1.5)
plt.text(873 + 5, 5, '873', rotation=90, verticalalignment='top', color='black')

# Configuraciones de ejes y diseño
plt.xlabel('Longitud de onda')
plt.ylabel('ABS')
plt.title('Barrido de Espectrofotometría')
plt.xlim(190, 1100)
plt.ylim(0, 5.5)

# Añadir leyenda
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```

### Extracción de Absorbancia en Longitudes de Onda Específicas
La tablita que se muestra en la imagen representa los valores exactos de absorbancia (ABS) para cada muestra en dos longitudes de onda de interés (en este caso 417/418 nm y 872/873 nm), las cuales corresponden a las líneas verticales trazadas arriba.

```python
# Extraer las filas donde la longitud de onda es más cercana a 417 y 873 nm
w1 = 417
w2 = 873

# Obtener el registro más cercano a w1 y w2
row_w1 = df.iloc[(df['Wavelength(nm)'] - w1).abs().argsort()[:1]]
row_w2 = df.iloc[(df['Wavelength(nm)'] - w2).abs().argsort()[:1]]

# Construir la tabla resumen
resumen = pd.DataFrame({
    'Ejemplo': [col.replace('(Abs)', '').strip() for col in columns_to_plot],
    f'ABS({w1})': [row_w1[col].values[0] for col in columns_to_plot],
    f'ABS({w2})': [row_w2[col].values[0] for col in columns_to_plot]
})

# Mostrar la tabla
display(resumen)
```
